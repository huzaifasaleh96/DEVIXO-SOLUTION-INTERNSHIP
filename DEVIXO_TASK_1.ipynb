```python
import matplotlib.pyplot as plt
import seaborn as sns

# Select only numerical columns for correlation analysis
numerical_df = df.select_dtypes(include=['number'])

# --- 1. Correlation Matrix for Numerical Features ---
print("\n--- Correlation Matrix for Numerical Features ---")
correlation_matrix = numerical_df.corr()

plt.figure(figsize=(14, 10))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix of Numerical Features')
plt.show()

# --- 2. Correlations with the 'Converted' Target Variable ---
print("\n--- Correlation with 'Converted' (Target Variable) ---")
if 'Converted' in numerical_df.columns:
    converted_correlations = correlation_matrix['Converted'].sort_values(ascending=False)
    print(converted_correlations)
else:
    print("'Converted' column not found in numerical features for correlation analysis.")

# --- 3. Relationship between Categorical Features and Key Numerical/Target Features ---
print("\n--- Visualizing Relationships between Categorical and Numerical Features ---")

# Box plots for 'Purchase_Amount_PKR' by key categorical features
categorical_for_boxplot = ['Gender', 'City', 'Marketing_Channel', 'Campaign', 'Device', 'Traffic_Source']

plt.figure(figsize=(18, 20))
for i, col in enumerate(categorical_for_boxplot):
    plt.subplot(4, 2, i + 1)
    sns.boxplot(data=df, x='Purchase_Amount_PKR', y=col)
    plt.title(f'Purchase Amount by {col}')
    plt.xlabel('Purchase Amount (PKR)')
    plt.ylabel(col)
plt.tight_layout()
plt.show()

# Box plots for 'Income_PKR' by 'Converted' status
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x='Converted', y='Income_PKR')
plt.title('Income by Conversion Status')
plt.xlabel('Converted (0=No, 1=Yes)')
plt.ylabel('Income (PKR)')
plt.show()

# Box plots for 'Purchase_Amount_PKR' by 'Converted' status
plt.figure(figsize=(8, 6))
sns.boxplot(data=df, x='Converted', y='Purchase_Amount_PKR')
plt.title('Purchase Amount by Conversion Status')
plt.xlabel('Converted (0=No, 1=Yes)')
plt.ylabel('Purchase Amount (PKR)')
plt.show()
```

# STEP 9 – 15 Business Insights

The following insights are generated from the actual marketing dataset used in this notebook.
Each insight is calculated from the dataset rather than being manually invented. The results
should be reviewed alongside the corresponding analysis and visualizations before submission.

**Important:** These are descriptive findings from this dataset. They show patterns/associations
and should not be interpreted as proof of causation.


```python
# STEP 9 – Generate 15 Evidence-Based Business Insights

# Helper function for readable percentage formatting
def pct(x):
    return f"{x * 100:.2f}%"

overall_conversion = df["Converted"].mean()

channel = df.groupby("Marketing_Channel").agg(
    conversion_rate=("Converted", "mean"),
    avg_purchase=("Purchase_Amount_PKR", "mean"),
    customers=("Converted", "size")
).sort_values("conversion_rate", ascending=False)

campaign = df.groupby("Campaign").agg(
    conversion_rate=("Converted", "mean"),
    avg_purchase=("Purchase_Amount_PKR", "mean"),
    customers=("Converted", "size")
).sort_values("conversion_rate", ascending=False)

city = df.groupby("City").agg(
    conversion_rate=("Converted", "mean"),
    avg_purchase=("Purchase_Amount_PKR", "mean"),
    customers=("Converted", "size")
).sort_values("conversion_rate", ascending=False)

device = df.groupby("Device").agg(
    conversion_rate=("Converted", "mean"),
    avg_purchase=("Purchase_Amount_PKR", "mean"),
    customers=("Converted", "size")
).sort_values("conversion_rate", ascending=False)

traffic = df.groupby("Traffic_Source").agg(
    conversion_rate=("Converted", "mean"),
    avg_purchase=("Purchase_Amount_PKR", "mean"),
    customers=("Converted", "size")
).sort_values("conversion_rate", ascending=False)

gender = df.groupby("Gender").agg(
    conversion_rate=("Converted", "mean"),
    avg_purchase=("Purchase_Amount_PKR", "mean"),
    customers=("Converted", "size")
).sort_values("conversion_rate", ascending=False)

email = df.groupby("Email_Opened").agg(
    conversion_rate=("Converted", "mean"),
    avg_purchase=("Purchase_Amount_PKR", "mean"),
    customers=("Converted", "size")
)

converted_stats = df.groupby("Converted")[
    ["Age", "Income_PKR", "Purchase_Amount_PKR",
     "Ad_Impressions", "Ad_Clicks", "Website_Visits",
     "Previous_Purchases", "Customer_Satisfaction"]
].mean()

correlations = df.select_dtypes(include="number").corr()["Converted"].drop("Converted")
correlations = correlations.sort_values(ascending=False)

# 15 evidence-based insights
insights = [
    f"1. Overall conversion: {df['Converted'].sum()} out of {len(df)} customers converted, giving an overall conversion rate of {pct(overall_conversion)}.",

    f"2. Marketing channel performance: {channel.index[0]} has the highest conversion rate at {pct(channel.iloc[0]['conversion_rate'])}, while {channel.index[-1]} has the lowest at {pct(channel.iloc[-1]['conversion_rate'])}.",

    f"3. Highest average purchase by channel: {channel['avg_purchase'].idxmax()} records the highest average purchase amount at PKR {channel['avg_purchase'].max():,.2f}.",

    f"4. Campaign performance: {campaign.index[0]} has the highest conversion rate at {pct(campaign.iloc[0]['conversion_rate'])}, while {campaign.index[-1]} has the lowest at {pct(campaign.iloc[-1]['conversion_rate'])}.",

    f"5. Highest average purchase by campaign: {campaign['avg_purchase'].idxmax()} has the highest average purchase amount at PKR {campaign['avg_purchase'].max():,.2f}.",

    f"6. City performance: {city.index[0]} has the highest conversion rate at {pct(city.iloc[0]['conversion_rate'])}, while {city.index[-1]} has the lowest at {pct(city.iloc[-1]['conversion_rate'])}.",

    f"7. Device performance: {device.index[0]} users have the highest conversion rate at {pct(device.iloc[0]['conversion_rate'])}, compared with {pct(device.iloc[-1]['conversion_rate'])} for {device.index[-1]} users.",

    f"8. Traffic-source performance: {traffic.index[0]} has the highest conversion rate at {pct(traffic.iloc[0]['conversion_rate'])}, while {traffic.index[-1]} has the lowest at {pct(traffic.iloc[-1]['conversion_rate'])}.",

    f"9. Gender comparison: {gender.index[0]} customers have a conversion rate of {pct(gender.iloc[0]['conversion_rate'])}, compared with {pct(gender.iloc[-1]['conversion_rate'])} for {gender.index[-1]} customers.",

    f"10. Income and conversion: converted customers have an average income of PKR {converted_stats.loc[1, 'Income_PKR']:,.2f}, compared with PKR {converted_stats.loc[0, 'Income_PKR']:,.2f} for non-converted customers.",

    f"11. Purchase value and conversion: converted customers have an average purchase amount of PKR {converted_stats.loc[1, 'Purchase_Amount_PKR']:,.2f}, compared with PKR {converted_stats.loc[0, 'Purchase_Amount_PKR']:,.2f} for non-converted customers.",

    f"12. Website engagement: converted customers average {converted_stats.loc[1, 'Website_Visits']:.2f} website visits, compared with {converted_stats.loc[0, 'Website_Visits']:.2f} among non-converted customers.",

    f"13. Previous purchase behavior: converted customers average {converted_stats.loc[1, 'Previous_Purchases']:.2f} previous purchases, compared with {converted_stats.loc[0, 'Previous_Purchases']:.2f} for non-converted customers.",

    f"14. Email engagement: customers who opened emails have a conversion rate of {pct(email.loc[1, 'conversion_rate'])}, while customers who did not open emails have a conversion rate of {pct(email.loc[0, 'conversion_rate'])}.",

    f"15. Correlation finding: the numerical feature with the strongest correlation with Converted (excluding the target itself) is {correlations.index[0]}, with a correlation of {correlations.iloc[0]:.3f}. This is an association in this dataset, not proof of causation."
]

print("=" * 90)
print("15 BUSINESS INSIGHTS – BASED ON THE ACTUAL DATASET")
print("=" * 90)

for insight in insights:
    print("\n" + insight)

print("\n" + "=" * 90)
print("Supporting summary tables")
print("=" * 90)

print("\nMarketing Channel Summary:")
display(channel.round(3))

print("\nCampaign Summary:")
display(campaign.round(3))

print("\nCity Summary:")
display(city.round(3))

print("\nDevice Summary:")
display(device.round(3))

print("\nTraffic Source Summary:")
display(traffic.round(3))

print("\nConverted vs Non-Converted Numerical Averages:")
display(converted_stats.round(2))
```
